```python
#Numpy Array Operations:
```

```python

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python

print("Shape:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)
```

```python

temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
```

```python

print("Maximum Temperature:", np.max(temperatures_w1))

print("Minimum Temperature:", np.min(temperatures_w1))

print("Mean Temperature:", np.mean(temperatures_w1))
```

```python


print("First three days temperatures:", temperatures_w1[:3])

print("Weekend temperatures:", temperatures_w1[-2:])

print("Middle three days temperatures:", temperatures_w1[2:5])
```

```python

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures for both weeks:")
print(temperatures)
```

```python

print("Shape of the array:", temperatures.shape)

print("Data type:", temperatures.dtype)

print("Total number of elements:", temperatures.size)

print("\nWeek 1 temperatures:", temperatures[0])

print("Week 2 temperatures:", temperatures[1])

print("Weekend temperatures for both weeks:")
print(temperatures[:, -2:])
```

```python
#Pandas Series:
```

```python

import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python

print("Mark of 1st rank student:", marks[0])

print("\nMarks of top 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\nMark of 3rd rank student:", marks.iloc[2])

print("\nStudents with marks greater than 90:")
print(marks[marks > 90])
```

```python

marks['Rank1'] = 100

print("Updated student marks:")
print(marks)

marks = marks.drop('Rank5')

print("\nMarks after removing Rank5:")
print(marks)

cgpa = marks / 10

print("\nCGPA of students:")
print(cgpa)
```

```python
#Pandas DataFrame:
```

```python

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("Transaction Data:")
print(transactions)
```

```python

print("1. Complete DataFrame:")
print(transactions)

print("\n2. First 5 rows:")
print(transactions.head())

print("\n3. Last 5 rows:")
print(transactions.tail())

print("\n4. Shape of DataFrame:")
print(transactions.shape)

print("\n5. Column Names:")
print(transactions.columns)

print("\n6. Data Types:")
print(transactions.dtypes)

print("\n7. ProductCategory and Amount columns:")
print(transactions[['ProductCategory', 'Amount']])

print("\n8. Last 3 columns:")
print(transactions.iloc[:, -3:])

print("\n9. Transactions from North region with Amount greater than 200:")
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

print("\n10. Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())

print("\n11. Unique Regions:")
print(transactions['Region'].unique())

print("\n12. Average Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("1. Updated Amount:")
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10

print("\n2. DataFrame after adding Discount:")
print(transactions)

transactions = transactions[transactions['TransactionID'] != 109]

print("\n3. DataFrame after removing TransactionID 109:")
print(transactions)

transactions = transactions.drop(columns=['Discount'])

print("\n4. Final DataFrame after deleting Discount column:")
print(transactions)
```
